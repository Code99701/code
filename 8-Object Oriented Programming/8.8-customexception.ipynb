#### Custom Exception (Raise and throw an  exception)

```python
class Error(Exception):
    pass

class dobException(Error):
    pass
```

```python
year = int(input("Enter your year of birth: "))
age = 2024 - year

try: 
    if age <= 30 and age >=0:
        print("the is valid so you can apply for the exam")
    else:
        raise dobException
except dobException:
    print("the is invalid so you cannot apply for the exam")
```
