##### Inheritance  

Inheritance is a fundamental concept in Object-Oriented Programming (OOP) that allows a class to inherit attributes and methods from another class. This lesson covers single inheritance and multiple inheritance, demonstrating how to create and use them in Python.

```python
## Inheritance(single inheritance)
## parent class 
class car:
    def __init__(self,windows,doors,enginetype):
        self.windows = windows
        self.doors = doors
        self.engine_type = enginetype 

    def drive(self):
        print(f"the person will drive the {self.engine_type} car")  

    
```

```python
car1 = car(4,5,"petrol")
print(car1.drive())
print(car1.windows)
print(car1.doors)   
```

```python
class Tesla(car):
    def __init__(self,windows,doors,enginetype,autopilot):
        super().__init__(windows,doors,enginetype)
        self.autopilot = autopilot

    def self_driving(self):
        print(f"Tesla supports {self.autopilot}")
```

```python
tesla = Tesla(4,5,"electric","full self driving")
print(tesla.drive())
print(tesla.self_driving())
```

```python
### Multiple inheritance
## when a class inherits from more than one parent class, it is called multiple inheritance.
## the child class inherits the properties and methods of all the parent classes.

# parent class 1
class animal:
    def __init__(self, name):
        self.name = name
    
    def speak(self):
        print("Subclass must implement this method")

# parent class 2
class pet:
    def __init__(self, owner):
        self.owner = owner
    
# child class
class dog(animal, pet):
    def __init__(self, name, owner, breed):
        animal.__init__(self, name)
        pet.__init__(self, owner)
        self.breed = breed
    
    def speak(self):
        return f"{self.name} says Woof!"
    
dog1 = dog("Buddy", "Alice", "Golden Retriever")
print(dog1.speak())
print(f"{dog1.name} is owned by {dog1.owner} and is a {dog1.breed}.")
```
