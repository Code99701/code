```python
import numpy as np

lst_marks = [23,43,56,87,34,92,61,48,17,94,21,83,57,49.60]
minimum,Q1,median,Q3,maximum = np.quantile(lst_marks,[0,0.25,0.5,0.75,1.0])
```

```python
minimum,Q1,median,Q3,maximum
```

```python
IQR = Q3 - Q1 
print(IQR)
```

```python
lower_fence = Q1 - 1.5*(IQR)
higher_fence = Q1 + 1.5*(IQR)
```
