```python
## Arithmetic operation

a = 10
b = 5
add_result = a + b
sub_result = a - b
mult_result = a * b
div_result = a/b
floor_result = a//b
modulus_result = a%b

expo_result = a**b

print(add_result)
print(sub_result)
print(mult_result)
print(div_result)
print(floor_result)
print(modulus_result)
print(expo_result)
```
