```python
# Syntax
lambda arguments: expression
```

```python
def addition(a,b):
    return a + b
```

```python
addition(2,3)
```

```python
addition = lambda a,b: a+b
type(addition)
print(addition(2,3))
```

```python
def even(num):
    if num % 2 == 0:
        return True
    else:
        return False
    
even(234)
```

```python
even1 = lambda num:num%2 == 0
even1(72134)
```

```python
def addition(x,y,z):
    return x+y+z

addition(2,3,4)
```

```python
addition1 = lambda x,y,z:x+y+z
addition1(12,13,14)
```

```python
## map() - It applies a function to all the items in an iterable and returns a map object (which is an iterator).
numbers=[1,2,3,4,5]
def square(num):
    return num**2

square(12)
```

```python
list(map(lambda x:x**2, numbers))
```
