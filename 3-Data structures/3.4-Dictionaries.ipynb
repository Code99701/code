## Dictionaries are unordered collection of items. they store data in key-value pairs.key must be unique and immutable (e.g., strings, numbers, or tuples), while values can be of any type

```python
## creating Dictionaries
empty_dict={}
print(type(empty_dict))
```

```python
empty_dict = dict()
empty_dict
```

```python
student = {"name":"Arnav","age":19,"grade":32}
print(student)
print(type(student))
```

```python
# error
student = {"name":"Arnav","age":19,"name":32}
print(student)
```

```python
## accessing Dictionary Elements
student = {"name":"Arnav","age":19,"grade":'A'}
print(student)
```

```python
print(student['grade'])
print(student['age'])

## accessing using get() method
print(student.get('grade'))
print(student.get('last_name'))
print(student.get('last_name',"Not Available"))
```

```python
## Modifying Dictionary elements
## Dictionary are mutable,so you can add, update or delete elements
print(student)
```

```python
student["age"] = 20 ##update value for the key
print(student)
student["address"]="India"  ## added a new key and value
print(student)
```

```python
del student["grade"] ## delete key and value pair

print(student)
```

```python
## Dictionary methods

keys = student.keys() ## get all the keys
print(keys)
values = student.values() ## get all values
print(values)

items = student.items()## get all the key vlaue pairs
print(items)
```

```python
## shallow copy
student_copy = student
print(student)
print(student_copy)
```

```python
student["name"] = "Arnav2"
print(student)
print(student_copy)
```

```python
student_copy1 = student.copy() ## shallow copy
print(student_copy1)
print(student)
```

```python
student["name"] = "Arnav3"
print(student_copy1)
print(student)
```

```python
### Iterating over Dictionaries
## you can use loops to iterate over dictionaries, keys,values, or items

## Iterating over keys
for keys in student.keys():
    print(keys)
```

```python
## Iterate over values
for value in student.values():
    print(value)
```

```python
## Iterate over key value pairs
for key,value in student.items():
   print(f"{key}:{value}")
```

```python
## Nested Dictionaries
students = {
    "student1":{"name":"Arnav","age:":19},
    "student2":{"name":"Peter","age:":28}
}
print(students)
```

```python
## Access Nested dictionaries elements
print(students["student1"]["name"])
print(students["student2"]["name"])
```

```python
students.items()
```

```python
## Iterating over the nested dicrionaries
for student_id,student_info in students.items():
    print(f"{student_id}:{student_info}")
    for key,value in student_info.items():
        print(f"{key}:{value}")
```

```python
## Dictionary Comprehension
Squares = {x:x**2 for x in range(5)}
print(Squares)
```

```python
## conditional dictionary comprehension
evens = {x:x**2 for x in range(10) if x%2 ==0}
print(evens)
```

```python
## Practical examples

## use a dictionary to count the frequency of elements in list

numbers = [1,2,2,3,3,3,4,4,4,4]
frequency = {}

for number in numbers:
    if number in frequency:
        frequency[number] += 1
    else:
        frequency[number] = 1
print(frequency)
```

```python
## Merge 2 Dictionaries into one

dict1 = {"a":1,"b":2}
dict2 = {"b":3,"c":4}
merged_dict = {**dict1,**dict2}
print(merged_dict)
```
