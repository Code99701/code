# Algerian Forest Fires Dataset

## Data Set Information

The dataset includes **244 instances** that regroup data of two regions of Algeria, namely:

- The **Bejaia** region located in the northeast of Algeria.
- The **Sidi Bel-Abbes** region located in the northwest of Algeria.

There are **122 instances for each region**.

The period covered is from **June 2012 to September 2012**.

The dataset includes:

- **11 attributes**
- **1 output attribute (class)**

The **244 instances** have been classified into:

- **Fire:** 138 instances
- **Not Fire:** 106 instances

## Attribute Information

1. **Date** : (DD/MM/YYYY) Day, month ('June' to 'September'), year (2012) Weather data observations.

2. **Temp** : Temperature noon (maximum temperature) in Celsius degrees: **22 to 42**

3. **RH** : Relative Humidity in %: **21 to 90**

4. **Ws** : Wind speed in km/h: **6 to 29**

5. **Rain** : Total day rain in mm: **0 to 16.8**

### FWI Components

6. **Fine Fuel Moisture Code (FFMC)** index from the FWI system: **28.6 to 92.5**

7. **Duff Moisture Code (DMC)** index from the FWI system: **1.1 to 65.9**

8. **Drought Code (DC)** index from the FWI system: **7 to 220.4**

9. **Initial Spread Index (ISI)** index from the FWI system: **0 to 18.5**

10. **Buildup Index (BUI)** index from the FWI system: **1.1 to 68**

11. **Fire Weather Index (FWI)** index: **0 to 31.1**

12. **Classes** : Two classes, namely **Fire** and **Not Fire**

## DATA CLEANING


```python
## missing values
dataset[dataset.isnull().any(axis=1)]
```

1. The dataset is converted into two sets based on Region from the 122th index. We can make a new column based on the Region.

   1 : "Bejaia Region Dataset"

   2 : "Sidi-Bel Abbes Region Dataset"

2. Add a new column with Region.

```python
dataset.loc[:122,'region'] = 0
dataset.loc[122:,'region'] = 1
df = dataset
```

```python
df[['region']] = df[['region']].astype(int)
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
## Removing the null values
df = df.dropna().reset_index()
```

```python
df.head()
```

```python
df.isnull().sum()
```

```python
df.iloc[[122]]
```

```python
## Remove the 122nd roww
df = df.drop(122).reset_index(drop=True)
```

```python
df.iloc[[122]]
```

```python
df.columns
```

```python
## Fix spaces in the column names
df.columns = df.columns.str.strip()
df.columns
```

```python
df.info()
```

### change the required columns as integrer type

```python
df.columns
```

```python
df[['month','day','year','Temperature','RH','Ws']] = df[['month','day','year','Temperature','RH','Ws']].astype(int)
```

```python
df.info()
```

### changing the other columns to float datatype

```python
objects = [features for features in df.columns if df[features].dtypes == 'O']
```

```python
for i in objects:
    if i != 'Classes':
        df[i]= df[i].astype(float)
```
